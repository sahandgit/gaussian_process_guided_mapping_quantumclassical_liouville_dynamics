# 02 — Running full-density `P0` cases

These are command templates for research runs. Run them in a terminal from the repository root after installation.

## Quick non-scientific debug run

```powershell
gp-mqcld-run --density_mode full --sampling_mode focused --P0 10 --R0 -15 --sigma_R 1.0 --dt 0.1 --n_steps 2 --n_train 64 --n_opt_steps 2 --snapshot_every 1 --panel_every 0 --skip_figures --quiet --out ".\runs\debug_P0_10"
```

## Production-style `P0=10`

