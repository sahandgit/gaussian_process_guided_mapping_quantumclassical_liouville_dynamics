# 03 — Loading results and basic diagnostics

This notebook shows the expected structure of saved runs and how to inspect diagnostics. Run a small/debug simulation first if production results are not available.

```python
from pathlib import Path
import numpy as np
import json

run_dir = Path("runs/P0_10")
print("Exists:", run_dir.exists())
print("Expected files:")
for name in ["pbme.npz", "pbme.json", "midpoint.npz", "midpoint.json"]:
    print("  ", run_dir / name)
```

```python
# Load arrays if files exist.
if (run_dir / "midpoint.npz").exists():
    data = np.load(run_dir / "midpoint.npz", allow_pickle=True)
    print("Keys:", list(data.keys())[:30])
    if "t" in data:
        print("Number of recorded times:", len(data["t"]))
else:
    print("Run the P0=10 command first, or change run_dir to an existing run.")
```

```python
# Simple diagnostic plot, if arrays are present.
import matplotlib.pyplot as plt
if (run_dir / "midpoint.npz").exists():
    data = np.load(run_dir / "midpoint.npz", allow_pickle=True)
    keys = set(data.keys())
    if {"t", "P0", "P1"}.issubset(keys):
        plt.plot(data["t"], data["P0"], label="P0")
        plt.plot(data["t"], data["P1"], label="P1")
        plt.xlabel("t")
        plt.ylabel("population")
        plt.legend()
        plt.show()
    else:
        print("Population keys not found. Available keys include:", sorted(list(keys))[:40])
```
